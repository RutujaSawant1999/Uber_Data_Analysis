## Data Preprocessing

```python
dataset['PURPOSE'].fillna("NOT",inplace = True)
```

```python
dataset.head()
```

```python
dataset['START_DATE'] = pd.to_datetime(dataset['START_DATE'], errors = 'coerce')

dataset['END_DATE'] = pd.to_datetime(dataset['END_DATE'], errors = 'coerce')
```

```python
dataset.info()
```

```python
from datetime import datetime

dataset['date'] =pd.DatetimeIndex(dataset['START_DATE']).date
dataset['time'] =pd.DatetimeIndex(dataset['START_DATE']).hour
```

```python
dataset.head()
```

```python
dataset['day-night'] =pd.cut(x=dataset['time'],bins = [0,10,15,19,24],labels = ['Morning', 'Afternoon', 'Evening', 'Night'])
```

```python
dataset.head()
```

```python
dataset.dropna(inplace = True)
```

```python
dataset.shape
```

### 4. In which months do people book Uber rides less frequently ?

```python
dataset.head()
```

#### Analyze Month

```python
dataset['MONTH'] = pd.DatetimeIndex(dataset['START_DATE']).month   # Extract month from START_DATE

month_label = {1.0 : 'Jan' , 2.0 : 'Feb', 3.0 : 'Mar', 4.0 : 'April', 5.0 : 'May', 6.0 :'June', 7.0 : 'July', 8.0 : 'Aug', 9.0 : 'Sep', 
               10.0 : 'Oct', 11.0 : 'Nov', 12.0 : 'Dec'}  # Assign string names to months

dataset["MONTH"] = dataset.MONTH.map(month_label)   # replace number month by string names

mon = dataset.MONTH.value_counts(sort=False)   # calculate count of every month
```

```python
dataset.head()
```

```python
df = pd.DataFrame({
    "MONTH": mon.values, # every month total count
    "VALUE COUNT": dataset.groupby('MONTH', sort=False)['MILES'].max()  # every month max miles value
})

p = sns.lineplot(data = df) # make line plot
p.set(xlabel="MONTHS", ylabel="VALUE COUNT")  # set axis labels
    
```

## Conclusion : January, November, December months people book uber rides less frequently.

### 5. On which days of the week do people book Uber rides the most ? 

```python
dataset['DAY'] = dataset.START_DATE.dt.weekday

day_label = {
    0: 'Mon' , 1: 'Tues', 2: 'Wed', 3: 'Thur', 4 : 'Fri', 5: 'Sat' , 6: 'Sun'}

dataset['DAY'] = dataset['DAY'].map(day_label)
```

```python
dataset.head()
```

```python
day_label = dataset.DAY.value_counts()

sns.barplot(x=day_label.index, y=day_label)
plt.xlabel('DAY')
plt.ylabel('COUNT')
```

## Conclusion: On Friday people book Uber rides the most.
